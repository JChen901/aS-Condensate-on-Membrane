# NAC Contact Definition Figures

Run this notebook from the `fig` folder to regenerate the six NAC-only PNG figures for the revised chain selections. The plotting implementation is kept in `plot_nac_contact_definition_figures.py` in this same folder, so this notebook stays as a small launcher.


## Setup

This cell finds `Analysis/NAC_contact_survival` whether Jupyter starts from the project root, the analysis folder, or this `fig` folder.


```python
from pathlib import Path
import importlib.util


def looks_like_analysis_dir(candidate):
    required = [
        "NAC_contact_relaxation_time",
        "NAC_antisymmetric_contact_relaxation_time",
        "NAC_banded_symmetric_contact_relaxation_time",
        "NAC_banded_antisymmetric_contact_relaxation_time",
        "fig",
    ]
    return all((candidate / name).exists() for name in required) and (candidate / "fig" / "plot_nac_contact_definition_figures.py").exists()


def find_analysis_dir():
    cwd = Path.cwd().resolve()
    starts = [
        cwd,
        cwd.parent,
        cwd / "Analysis" / "NAC_contact_survival",
        cwd / "NAC_contact_survival",
    ]
    seen = set()
    for start in starts:
        if start in seen:
            continue
        seen.add(start)
        for candidate in (start, *start.parents):
            if looks_like_analysis_dir(candidate):
                return candidate
    raise FileNotFoundError(
        "Could not find Analysis/NAC_contact_survival. "
        "Start Jupyter from the project root, the analysis folder, or the fig folder."
    )


ANALYSIS_DIR = find_analysis_dir()
FIG_DIR = ANALYSIS_DIR / "fig"
PLOT_SCRIPT = FIG_DIR / "plot_nac_contact_definition_figures.py"

print(f"Analysis folder: {ANALYSIS_DIR}")
print(f"Figure folder:   {FIG_DIR}")
print(f"Plot script:     {PLOT_SCRIPT}")
```

## Check Inputs

The figures use summary CSVs for the bars and per-trajectory CSVs for the hollow replicate circles. If you have deleted the CSV files, rerun the four analysis notebooks first, then run this notebook.


```python
VARIANT_FOLDERS = [
    "NAC_contact_relaxation_time",
    "NAC_antisymmetric_contact_relaxation_time",
    "NAC_banded_symmetric_contact_relaxation_time",
    "NAC_banded_antisymmetric_contact_relaxation_time",
]

required_metric_csvs = []
for folder in VARIANT_FOLDERS:
    base = ANALYSIS_DIR / folder
    required_metric_csvs.extend(
        [
            base / "nac_contact_outputs" / "nac_contact_survival_t_01_summary.csv",
            base / "nac_contact_outputs" / "nac_contact_survival_t_1e_summary.csv",
            base / "nac_contact_acf_outputs" / "nac_contact_acf_t_01_summary.csv",
            base / "nac_contact_acf_outputs" / "nac_contact_acf_t_1e_summary.csv",
        ]
    )

missing = [path for path in required_metric_csvs if not path.exists()]
if missing:
    message = "Missing required NAC CSV files:" + "\n" + "\n".join(
        f"- {p.relative_to(ANALYSIS_DIR)}" for p in missing
    )
    raise FileNotFoundError(message)

print(f"Found {len(required_metric_csvs)} required metric CSV files.")
```

## Generate Figures

This imports the original plotting module and runs `main()`. Existing PNGs in `fig/` with the same names will be overwritten.


```python
spec = importlib.util.spec_from_file_location("nac_contact_definition_plotter", PLOT_SCRIPT)
plotter = importlib.util.module_from_spec(spec)
spec.loader.exec_module(plotter)

plotter.BASE_DIR = ANALYSIS_DIR
plotter.FIG_DIR = FIG_DIR
plotter.main()

print("Finished regenerating NAC contact definition figures.")
```
